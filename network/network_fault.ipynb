```python
import numpy as np
import shap

explainer = shap.TreeExplainer(model_xgb)

def get_severity(probability):
    if probability < 30:
        return "Low"
    elif probability < 60:
        return "Medium"
    elif probability < 80:
        return "High"
    return "Critical"

def get_root_cause(data):
    shap_values = explainer.shap_values(data)[0]
    top = np.argsort(np.abs(shap_values))[-3:][::-1]

    return [data.columns[i] for i in top]
```

```python
def predict_network_fault(data):
    prediction = model_xgb.predict(data)[0]
    probability = model_xgb.predict_proba(data)[0][1] * 100

    result = {
        "fault": int(prediction),
        "fault_probability": round(float(probability), 2),
        "severity": get_severity(probability)
    }

    if prediction == 1:
        fault_type = le.inverse_transform(
            [fault_model.predict(data)[0]]
        )[0]

        result["fault_type"] = fault_type
        result["top_features"] = get_root_cause(data)

    return result
```
